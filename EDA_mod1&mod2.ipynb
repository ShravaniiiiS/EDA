```python
#23
import numpy as np
dtype = [
    ('title', 'U30'),
    ('author', 'U20'),
    ('price', 'f4')
]

books = np.array([
    ('Python Basics', 'John Smith', 450.0),
    ('Data Science', 'Alice Brown', 550.0),
    ('Machine Learning', 'David Lee', 650.0),
    ('NumPy Guide', 'Robert Green', 350.0)
], dtype=dtype)
print(books)

#24
print("All Book Titles:")
print(books['title'])

print("Second Book:")
print(books[1])

amount = 500

print("Books priced below", amount, ":")
print(books['title'][books['price'] < amount])

```

```python
#25
import numpy as np

dtype_dict = {
    'names': ['title', 'author', 'price'],
    'formats': ['U30', 'U20', 'f4']
}

dtype_list = [
    ('title', 'U30'),
    ('author', 'U20'),
    ('price', 'f4')
]

dt1 = np.dtype(dtype_dict)
dt2 = np.dtype(dtype_list)

print("Dictionary form:")
print(dt1)

print("\nList-of-tuples form:")
print(dt2)

print("\nAre both dtypes equivalent?")
print(dt1 == dt2)
```

```python
#26
import numpy as np

dtype = [
    ('title', 'U30'),
    ('author', 'U20'),
    ('price', 'f4')
]

books = np.array([
    ('Python Basics', 'John Smith', 450.0),
    ('Data Science', 'Alice Brown', 550.0),
    ('Machine Learning', 'David Lee', 650.0),
    ('NumPy Guide', 'Robert Green', 350.0)
], dtype=dtype)
data_rec = books.view(np.recarray)

print("Book prices:")
print(data_rec.price)
```

```python
#27
import numpy as np

dtype = [
    ('roll_no', 'i4'),
    ('name', 'U20'),
    ('marks', 'f4')
]

students = np.array([
    (101, 'Rahul', 78.5),
    (102, 'Ananya', 92.0),
    (103, 'Kiran', 85.5),
    (104, 'Sneha', 95.0),
    (105, 'Arjun', 70.5)
], dtype=dtype)

sorted_students = np.sort(students, order='marks')

sorted_students = sorted_students[::-1]

print("Students sorted by marks in descending order:")
print(sorted_students)
```
