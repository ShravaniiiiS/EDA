```python
import numpy as np
import pandas as pd

index = pd.MultiIndex.from_product(
    [[2013, 2014], [1, 2]], names=["year", "visit"]
)

columns = pd.MultiIndex.from_product(
    [["Bob", "Guido", "Sue"], ["HR", "Temp"]],
    names=["name", "type"]
)

data = np.round(np.random.randn(4, 6), 1)
health_data = pd.DataFrame(data, index=index, columns=columns)

print(health_data)

health_data['Guido']
health_data[('Guido', 'HR')]
health_data.iloc[:2, :2]
health_data.loc[:, ('Bob', 'HR')]

idx = pd.IndexSlice
health_data.loc[idx[:, 1], idx[:, 'HR']]

index = pd.MultiIndex.from_product([['a', 'c', 'b'], [1, 2]])
data = pd.Series(np.random.rand(6), index=index)
data.index.names = ['char', 'int']

try:
    data['a':'b']
except KeyError as e:
    print(type(e))
    print(e)

data = data.sort_index()
print(data['a':'b'])

print(health_data.groupby(level='year').mean())
print(health_data.groupby(level='year').mean().groupby(level='type', axis=1).mean())
```

```python
import pandas as pd
data = {
    'department': ['CSE', 'AIML', 'ECE', 'CSE', 'AIML', 'ECE',
                   'CSE', 'AIML', 'ECE', 'CSE', 'AIML', 'ECE'],
    'gender': ['M', 'F', 'M', 'F', 'M', 'F',
               'M', 'F', 'M', 'F', 'M', 'F'],
    'salary': [75000, 82000, 70000, 78000, 90000, 72000,
               76000, 85000, 69000, 80000, 88000, 71000],
    'age':[23,34,45,35,32,24,25,26,54,56,47,58]
}
df=pd.DataFrame(data)
#50 
average=df.pivot_table(values='salary',index='gender',columns='department',aggfunc='mean')

#51 
count=df.pivot_table(values='salary',index='gender',columns='department',aggfunc='count')

#52
margin=df.pivot_table(values='salary',index='gender',columns='department',aggfunc='min',margins=True)
print(margin)
print()
print(average)
print()
print(count)
print()
#53
df['age_category'] = pd.cut(df['age'], bins=[0,26,35,100], labels=['junior','mid','senior'], right=True)
pivot=df.pivot_table('salary',index='age_category',columns='department',aggfunc='mean')
print(pivot)
print()

#54
# 54
pivot2 = df.pivot_table(
    index='age_category',
    values=['salary', 'age'],
    aggfunc={'salary': 'sum', 'age': 'mean'}
)
print(pivot2)
```
